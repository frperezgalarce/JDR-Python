# Explicit R reference from a Python notebook

Version 2. R is invoked only via Rscript. The notebook reports a skip when R is unavailable rather than attempting R syntax in a Python kernel.

```python
from pathlib import Path
import sys
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'src' / 'jdr.py').exists())
if str(ROOT) not in sys.path: sys.path.insert(0, str(ROOT))
import numpy as np
import pandas as pd
```

```python
import shutil
from src.metrics import integrate_cross_spectrum_real_R, integrate_cross_spectrum_real
rng = np.random.default_rng(42)
t = np.sort(rng.uniform(0,3,40)); x = np.sin(2*np.pi*.7*t)
if shutil.which('Rscript'):
    for convention in ['standard','legacy']:
        r = integrate_cross_spectrum_real_R(.3,.8,t,x,t,x,convention=convention)
        py = integrate_cross_spectrum_real(.3,.8,t,x,t,x,convention=convention)
        assert np.isclose(r,py,rtol=1e-6,atol=1e-9)
    print('R references passed')
else:
    print('Rscript unavailable: reference check not executed')
```
